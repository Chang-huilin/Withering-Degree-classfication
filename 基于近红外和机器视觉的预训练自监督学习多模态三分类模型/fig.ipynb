```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score

# 混淆矩阵数据（真实标签 vs 预测标签）
confusion_matrix = np.array([
    [29, 2, 1],
    [1, 48, 1],
    [0, 3, 35]
])

# 类别标签
class_labels = ['Class 1', 'Class 2', 'Class 3']

# -------------------- 1. 绘制蓝色混淆矩阵 --------------------
plt.figure(figsize=(6, 6))
sns.set(font_scale=2)
ax = sns.heatmap(confusion_matrix, 
                 annot=True, 
                 fmt='d',
                 cmap='Blues',
                 cbar=True,
                 xticklabels=class_labels,
                 yticklabels=class_labels)
ax.set_title('Confusion Matrix (Blue Scale)', pad=20)
ax.set_xlabel('Predicted Class')
ax.set_ylabel('True Class')
plt.tight_layout()
plt.show()

# -------------------- 直接通过混淆矩阵计算准确率 --------------------
total_samples = np.sum(confusion_matrix)
correct_predictions = np.trace(confusion_matrix)  # 对角线元素之和
manual_accuracy = correct_predictions / total_samples
False_Negatives = np.sum(confusion_matrix, axis=1) - np.diag(confusion_matrix)
False_Positives = np.sum(confusion_matrix, axis=0) - np.diag(confusion_matrix)

print(f"\nManual Accuracy : {manual_accuracy:.4f}")
print(f"False Negatives: {False_Negatives}")
print(f"False Positives: {False_Positives}")
```
